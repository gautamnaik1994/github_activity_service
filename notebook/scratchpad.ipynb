```python
import requests
from collections import defaultdict, Counter
import json
```

```python

def fetch_events(username):
    url = f"https://api.github.com/users/{username}/events/public"
    resp = requests.get(url)
    resp.raise_for_status()
    return resp.json()


def summarize_activity(events, username):
    repos = defaultdict(list)
    for event in events:
        repo_name = event["repo"]["name"]
        event_type = event["type"]
        repos[repo_name].append(event_type)
    
    summary = []
    for repo, types in repos.items():
        counts = Counter(types)
        top3 = counts.most_common(3)
        owner = repo.split("/")[0]
        summary.append({
            "repo": repo,
            "is_owned_by_user": owner.lower() == username.lower(),
            "top_3_events": [e for e, _ in top3]
        })
    return summary
```

```python
# username = "ge0ffrey"
username = "gautamnaik1994"
events = fetch_events(username)
summary = summarize_activity(events, username)
print(json.dumps(summary, indent=2))
```
